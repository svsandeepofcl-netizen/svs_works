```python
class Solution(object):
    def moveZeroes(self, nums):
        """
        :type nums: List[int]
        :rtype: None Do not return anything, modify nums in-place instead.
        """

        trans_pos=0

        for i in range(len(nums)):

            if nums[i]!=0:
                nums[trans_pos]=nums[i]
                trans_pos+=1
            
        
        while trans_pos<len(nums):
            nums[trans_pos]=0
            trans_pos+=1
        
        return nums


obj=Solution()
print(obj.moveZeroes([0,1,0,3,12]))
```

```python
#brute force approach
class Solution(object):
    def mergeAlternately(self, word1, word2):
        """
        :type word1: str
        :type word2: str
        :rtype: str
        """
        if len(word1)>len(word2):
            loop_length=len(word2)

        elif len(word1)<len(word2):
            loop_length=len(word1)

        else:
            loop_length=len(word1)
        
        output=''

        for i in range(loop_length):
            output=output+word1[i]+word2[i]

        if len(word1)<len(word2):
            output=output+word2[i+1:len(word2)]
            return output
        
        output=output+word1[i+1:len(word1)]
        return output

obj=Solution()
word1 = "abd"
word2 = "pqrst"

print(obj.mergeAlternately(word1,word2))
```

```python
#brute force approach
class Solution(object):
    def mergeAlternately(self, word1, word2):
        """
        :type word1: str
        :type word2: str
        :rtype: str
        """
        loop_length=min(len(word1), len(word2))
        
        output=''

        for i in range(loop_length):
            output=output+word1[i]+word2[i]

        output=output+word1[loop_length:]
        output=output+word2[loop_length:]

        return output

obj=Solution()
word1 = "abcdfgh"
word2 = "pq"

print(obj.mergeAlternately(word1,word2))
```

```python
class Solution:
    def compress(self, chars: list[str]) -> int:

        left = 0
        write = 0

        while left < len(chars):

            right = left

            # move right until character changes
            while right < len(chars) and chars[right] == chars[left]:
                right += 1

            # count of current character
            count = right - left

            # write character
            chars[write] = chars[left]
            write += 1

            # write count if > 1
            if count > 1:
                for digit in str(count):
                    chars[write] = digit
                    write += 1

            # move to next group
            left = right

        return write
    

obj=Solution()
print(obj.compress(["a","a","b","b","c","c","c"]) )
```

```python
class Node:
    def __init__(self, key, value):
        self.key= key
        self.value=value
        self.prev= None
        self.next= None

class LRU:
    def __init__(self, capacity):
        self.capacity=capacity
        self.cache = {}

        #Intial head and tail
        self.left=Node(0,0)
        self.right=Node(0,0)

        #Linking Head and Tail
        self.left.next=self.right
        self.right.prev=self.left
        
    def remove(self, Node):
        # here we are delinking node so we are getting its previous and
        #  next values so that we can assign previous value as a prev value to next and vice versa
        prev=Node.prev
        nxt=Node.next

        prev.next=nxt
        nxt.prev=prev

    def insert(self, Node):

        prev=self.right.prev
        nxt=self.right

        #Here we are trying to add node at last before tail(self.right)
        prev.next=Node
        Node.prev=prev
        Node.next=nxt
        nxt.prev=Node


    def get(self, key):
        if key in self.cache:
            self.remove(Node)
            self.insert(Node)        
            return self.cache[key].value
        return -1
    

    def put(self, key, value):
        if key in self.cache:
            self.remove(Node)
        

        self.cache[key]=Node(key, value)
        self.insert(self.cache[key])

        if len(self.cache)>len(self.capacity):
            lru= self.left.next
            self.remove(lru)
            del self.cache[lru.key]
            
```
